```python
# 실습 2

회원 = pd.DataFrame({
    "이름": ["지민", "서준", "하린", "도윤"],
    "이메일": ["jimin@test.com", None, "harin@test.com", None],
    "연락처": [None, "010-2222-2222", "010-3333-3333", None]
})

회원.isnull()
print()
회원.isna()
print()
회원.notna()
```

```python
# 실습 3.

과제 = pd.DataFrame({
    "이름": ["민준", "서연", "지우", "수빈", "현우"],
    "점수": [90, None, 0, 85, None]
})

과제.isnull()
```
