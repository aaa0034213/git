```python
import pandas as pd

df = pd.DataFrame({
    "이름": ["철수", "영희", "민수"],
    "점수": [80, 90, 70],
    "출석": ["출석", "출석", "결석"]
})

print(df)
# df의 정보를 알고 싶을 때
df.info()
print()
# 전체 정보 말고 위에서 2개 혹은 위에서 3개 보고싶을 경우
# head(개수) - 표에서 위엣 개수만큼 데이터 확인
df.head(1)
print()

# df 행과 열의 개수를 알고 싶을 때
df.shape
print()

```

```python
# 빈 값 확인 기능 isnull()
df.isnull()

# 열마다 빈 값이 몇개인지 확인
df.isna().sum()
```

```python
# 점수가 비어있는 학생만 찾기
df["점수"].isna()
```

```python
# 점수가 있는 학생만 찾기
# 1. 열 선택
df["점수"].notna()
```

```python
df[df["점수"].notna()]
```

```python
df[df["점수"].isna()]
```

```python
# 범위로 데이터 가져오기

df

# 행의 번호를 이용해서 데이터 가져오기
df[0:3]     # 0부터 2까지
df[0:2]     # 0부터 1까지

# df 짜르기 원하는 부분
df.iloc[0:2]    # [시작범위:끝범위]

# 1번부터 끝까지 가져오기
df.iloc[1:]

# 행과 열 범위를 함께 선택할 수도 있다.
df.iloc[0:1, 0:1]

# 위에 두 개 차이?
# df[0:2] -> 행 범위로 가져온다
# df.iloc[] -> 행,열 한꺼번에 가져온다.

#-------------------------------------------------

# 이름표로 가져오기
df.loc[0:2]
```

```python
df.loc[0:1]
```

```python
df.iloc[0:1]
```

```python
# loc, iloc 차이점
# loc 끝번호 포함 o 
# iloc 는 끝번호 포함 x
```

```python
# 열 이름으로 각 데이터 가져오기.

df[['이름','출석']] 

df[['이름','출석']].head(2) 

# 파일들을 가져와서 데이터 읽다보면 첫 번째 행 대부분 컬럼명
# 첫 번째 행을 빼고 내가 원하는 컬럼들
# 하고 두 번째 줄부터 보고 싶을 때 사용하는 방법

df.loc[1:,['이름','출석']]

# 데이터 맨 끝에서 확인
df.tail(2)

# 행을 이용해서 끝에서 2개 가져오기
df.iloc[-2:]
```
